```typescript
// Truthy vs Falsy
/**
 * In JavaScript and TypeScript, values can be classified as "truthy" or "falsy" based on how they evaluate 
 * in a boolean context.
 * 
 * Falsy values include:
 * false, 0, -0, "" (empty string), null, undefined, NaN (Not-a-Number)
 * 
 * Truthy values include:
 * true, any non-zero number, any non-empty string ("0", "false", ""), arrays (even empty ones: []),
 * objects (even empty ones: {}), funcitons, symbols, BigInts (except On)
 */

if (0) {
    console.log("This won't print"); // 0 is falsy
}

if ("hello") {
    console.log("This will print"); // "hello" is truthy
}

if ([]) {
    console.log("This will print too"); // Empty array is truthy
}
```

```typescript
// Optional Chaining (?.)
/**
 * Optional chaining (ES11) allows you to safely access nested objects properties or call functions
 * without worrying if an intermediate property or function exists. If a property or function in the chain
 * is null or undefined, the expression short-circuits and evaluates to undefined instead of throwing an error.
 */

// object?.property;
// object>?.['property'];
// object?.method();

const user = {
    name: 'Alice',
    address: {
        street: '123 Main St'
    }
};

console.log(user.address.street); // Output: 123 Main St
console.log(user.address?.city); // Output: undefined (no error)
console.log(user.contact?.email); // Output: undefined (user.contact is undefined, no error)

const user2 = {};
console.log(user2.address?.street); // Output: undefined (user2.address is undefined, no error)
// Calling a method that might not exist
```

```typescript
// While loops
/* The while loop executes a block of code as long as 
* a specified condition is true
* It's important to ensure that the condition eventually
* becomes false to avoid infinite loops.
*/
// while (condition) {
    // code to be executed
// }

let i = 0;
while (i < 5) {
    console.log("The number is " + i);
    i++; // Increment i to eventuallyu make the condition
    // false and exit the loop
}
```

```typescript
// do...while loop
/**The do...while loop is similar to the while loop, but it guarantees that the code block is executed 
 * at least once, because the condition is checked *after* the block is executed.
 */

//do {
    // code to be executed
//} while (condition);

let i = 0;
do {
    console.log("The number is " + i);
    i++;
} while (i < 5)

// Example showing it runs at least once:
let j = 5;
do {
    console.log("This will print once: " + j);
    j++;
} while (j < 5); // Condition is initially false, but the block runs once
```

```typescript
// for loop
//* The for loop is commonly used when you know the number of times you want to loop through
// a block of code. It consists of three optional pairs: initializaiton, condition, and final expression (usually
// increment/decrement) */

//for (initializaiton; condition; finalExpression) {
    // code to be executed
//}

for (let i = 0; i < 5; i++) {
    console.log("The number is " + i);
}

```

```typescript
// for...of Loop
//** The for...of loop (ES6) iterates over the values of an iterable object (like Arrays, Strings,
// Maps, Sets, etc.). It's a simpler way to loop through elements compared to a traditional for loop
// with an index. */

//for (variable of iterable) {
    // code to be executed for each value
//}

const colors = ["red", "green", "blue"];
for (let color of colors) {
    console.log(color);
}

const greeting = "Hello";
for (let char of greeting) {
    console.log(char);
}
```

```typescript
// for..in loop
/** The for...in loop iterates over the enumerable properties (keys) of an object. It is gnerally not recommended for iterating
 * over Arrays because it can iterate over the index as strings and might include inherited properties.
 */

//for (variable in object) {
    // code to be executed for each property
//}

const person = { name: "Alice", age: 30, city: "New York" };
for (let key in person) {
    console.log(key + ": " + person[key]);
}

// Use for...of for iterating over array values and Object.keys(), Object.values(), or Object.entries() for iterating over
// object object keys/values/entries.
```

```typescript
// break and continue
/** You can control the flow of loops using break and continue staements:
 *  - break : Terminates the current loop entirely and continues execution at the first statement after the loop.
 *  - continue : Skips the rest of the current iteration of the loop and continues with the next iteration
 */

// break example
for (let i = 0; i < 10; i++) {
    if (i === 5) {
        break; // Exit the loop when i is 5
    }
    console.log("Break example: " + i);
}
// Output: 0, 1, 2, 3, 4

// continue example
for (let i = 0; i < 10; i++) {
    if (i % 2 === 0) {
        continue; // skip even numbers
    }
    console.log("Continue Example: " + i);
}
// Output: 1, 3, 5, 7, 9
```

```typescript
// Looping Over Arrays
/** There are several ways to loop over arrays in JavaScript: */

// Traditional for loop: Using an index.
const arr = [10, 20, 30];
for (let i = 0; i < arr.length; i++) {
    console.log(arr[i]); // Access element by index
}

// for...of loop (Recommended for values): Iterates directly over element values.
const arr = [10, 20, 30];
for (let values of arr) {
    console.log(values); // Access element value
}

// forEach() method: Executes a provided function once for each array element.
const arr = [10, 20, 30];
arr.forEach(function(value, index) {
    console.log(index + ": " + value);
});

// for...in loop (Not recommended for arrays): Iterates over the enumerable properties (indices) of the array.
const arr = [10, 20, 30];
for (let index in arr) {
    console.log(index + ": " + arr[index]); // Access element by index
}
```

```typescript
// Nested Loops
/** A nested loop is a loop inside another loop. The inner loop executes completely for each iteration of the 
 * outer loop.
 * Nested loops are often used when working with multi-dimensional arrays or when you need to perform operations
 * that involve combinations of elements from different sets.
 */

for (let i = 0; i < 3; i++) {
    console.log("Outer loop iteration: " + i);
    for (let j = 0; j < 2; j++) {
        console.log(" Inner loop iteration: " + j);
    }
}
// **Outer loop iteration: 0
// Inner loop iteration: 0
//  Inner loop iteration: 1
// Outer loop iteration: 1
//  Inner loop iteration: 0
//  Inner loop iteration: 1
// Outer loop iteration: 2
//  Inner loop iteration: 0
//  Inner loop iteration: 1
// */
```

```typescript
// Loop Performance Tips
/** While JavaScript engines are highly optimized, consider these tips for performance in loops,
 * especially for large datasets:
 * - Cache arrary length: Accessing `array.length` in every iteration can be slighly less
 * efficient than storing it in a variable before the loop.
 */

// Less efficient
for (let i >= 0; i < arr.length; i++) { ... }

// More efficient
const len = arr.length;
for (let i = 0; i < len; i++) { ... }

/** Avoid complex operations in the condition: Keep the loop condition simple. Choose the right
 * loop: `for...of` is generally good for iterating values, `forEach` is good for simple
 * iteration with a callback, traditional `for` is good for control over index. Avoid `for..in`
 * for arrays.
 * For most typical web development tasks, performance differences are negligible, but these tips
 * can be useful in performance-critical scenarios.
 */
```
