````typescript
// Default Parameters
/** Default parameters (ES6) allow you to specify default values for function parameters. If an argument is not provided for
 * a pararmeter when the funciton is called (or if the argument is `undefined`), the default value will be used.
 */

//function funcitonName(param1 = defaultValue1, param2 = defaultValue2, ...) {
    // code
//}

function greet(name = "Guest", greeting = "Hello") {
    console.log(`${greeting}, ${name}!`);
}

greet("Alice", "Hi"); // Output: Hi, Alice!
greet("Bob"); // Output: Hello, Bob! (greeting uses default)
greet(); // Output: Hello, Guest! (both use defautls)
greet(undefined, "Hey"); // Output: Hey, Guest! (name uses default because undefined)
greet(null, "Hey"); // Output: Hey, null! (null is a value, not undefined)
````

````typescript
// Rest Parameters
/** Rest parameters (ES6) allow a function to accept an indefinite number of arguments 
 * as an array. They are denoted by three dots (...) followed by a parameter name.
 * The rest parameter must be the last parameter in the function defintion
 */

//function funcitonName(param1, param2, ...restOfArgument) {
    // restOfArguments is an array
//}

function sumAll(...numbers) {
    let total = 0;
    for (let num of numbers) {
        total += num;
    }
    return total;
}

console.log(sumAll(1, 2, 3)); // Output: 6
console.log(sumAll(10, 20, 30, 40)); // Output: 100
console.log(sumAll()); // Output: 0
````

````typescript
// Callback Functions
/** A callback function is a function that is passed as an argument to another function
 * and is executed later, often after some asynchronous operatiuon or event has completed.
 * Callbacks are a fundamental concept in asynchronous JavaScript and event handling.
 */

// Example (Synchronous Callback):
function doOperation(x, y, callback) {
    const result = x + y;
    callback(result); // Call the callback funciton with the results
}

doOperation(5, 3, function(sum) {
    console.log("The sum is: " + sum); // This is the callback
}); // Output: The sum is 8

// Example (Asynchronous Callback - setTimeout):
function fetchData(callbacks) {
    // Simulate fetching data asynchronously
    setTimeout(function() {
        const data = { id: 1, name: 'Product A' };
        callbacks(data); // Call the callback with the fetched data
    }, 1000); // Wait for 1 second
}

console.log("Fetching data...");
fetchData(function(product) {
    console.log("Data received:", product); // This callback executes later
});
console.log("Request sent.");
// Output order might be:
// Fetching data...
// Request sent.
// Data received: { id: 1, name: 'Product A' }
````

````typescript
// Modifying Arrays
/** Arrays are mutable, meaning you can change their elements, add new elements, or remove elements.
 * - Changing elements: Assign a new value using the index.
 */

const colors = ["red", "green", "blue"];
colors[1] = "yellow";
console.log(colors); // Output: ["red", "yellow", "blue"]

/** Adding elements:
 * - Using `push()` to add to the end.
 * - Using `unshift()` to add to the beginning.
 * - Assigning to an index beyond the current length.
 * 
 * Removing Elements:
 * - Using `pop()` to remove from the end.
 * - Using `shift()` to remove from the beginning.
 * - Using `splice()` to remove from a specific index.
 */

// Example (Adding/Removing):

const fruits = ["apple", "banana"];
fruits.push("cherry"); // Add to end
console.log(fruits); // Output: ["apple", "banana", "cherry"]

fruits.unshift("strawberry"); // Add to beginning
console.log(fruits); // Output: ["strawberry", "apple", "banana", "cherry"]

let lastFruit = fruits.pop(); // Remove from end
console.log(lastFruit); // Output: cherry
console.log(fruits); // Output: ["strawberry", "apple", "banana"]

let firstFruit = fruits.shift(); // Remove from beginning
console.log(firstFruit); // Output: strawberry
console.log(fruits) // Output: ["apple", "banana"]

let spliceFruit = fruits.splice(1); // Splices banana
console.log(spliceFruit); // Output: "banana"
console.log(fruits) // Output: ["apple"]
````

````typescript
// Array Methods (push, pop, etc.)
/** JavaScript arrays have many built-in methods for common operators. We've already seen some for adding and removing
 * elements:
 * - `push()`: Adds one or more elements to the end and returns the new length.
 * - `pop()`: Removes the last element and returns that element.
 * - `unshift()`: Adds one or more elements to the beginning and returns the new length.
 * - `shift()`: Removes the first element and returns that element.
 * - `splice(startIndex, deleteCount, item1, ...)`: Adds/removes elements from a specific index. Returns an array of
 * removed elements.
 * - `slice(startIndex, endIndex): Returns a shallow copy of a portion of an array into a new array. Does not modify
 * the original array.
 * - `concat()`: Joins two or more array ands returns a new array.
 * - `indexOf()`: Returns the first index at which a given element can be found, or -1 if not present.
 * - `includes()` (ES7): Checks if an array contains a certain value, returns true or false.
 */

const numbers = [1, 2, 3, 4, 5];
const removed = numbers.splice(2, 1); // Remove 1 element starting at index 2
console.log(removed); // Output: [3]
console.log(numbers); // Output: [1, 2, 4, 5]

const sliced = numbers.slice(1, 3); // Get elements from index 1 up to (but not including) index 3
console.log(sliced); // Output: [2, 4]
console.log(numbers); // Original array is unchanged: [1, 2, 4, 5]
````

````typescript
// Iterating Arrays
/** Besides traditional loops (for, whjile, do...while, for...of), arrays have several built-in iteration methods that accept callback functions:
 * - `forEach(callback)`: Executes a callback for each element. Good for performing an action on each element.
 * - `map(callback)`: Creates a new array with the results of calling a callback on every element. Good for transforming an array.
 * - `filter(callback)`: Creates a new array with elements that pass a test implemented by the callback. Good for selecting elements.
 * - `reduce(callback, initialValue)`: Executes a reducer function on each element, resulting in a single output value. Good for aggregating data.
 * - `some(callback)`: Checking if at least one element in the array passes the test implemented by the callback. Returns true or false.
 * - `every(callback)`: Checks if all elements in the array pass the test implemented by the callback. Returns true or false.
 */

const numbers = [1, 2, 3, 4, 5];

numbers.forEach(num => console.log(num * 2)); // Output: 2, 4, 6, 8, 10

const doubled = numbers.map(num => num * 2);
console.log(doubled); // Output: [2, 4, 6, 8, 10]

const evens = numbers.filter(num => num % 2 == 0);
console.log(evens); // Output: [2, 4]
````

````typescript
// map, filter, reduce
/** These are three of the most powerful and commonly used array iteration methods, enabling functional programming patterns.
 * - `map(callback)`: Transforms each element. The callback should return the new value for the element.
 */

const numbers = [1, 2, 3];
const squared = numbers.map(num => num * num) 
console.log(squared)// [1, 4, 9]

// `filter(callback)`: Selects elements. The callback should return `true` to keep the element, `false` to discard it.

const ages = [16, 18, 20, 17];
const adults = ages.filter(age => age >= 18); 
console.log(adults) // [18, 20]

// `reduce(callback, initialValue)`: Reduces an array to a single value. The callback takes an accumulator and the current value.

const numbers1 = [1, 2, 3, 4];
const sum = numbers1.reduce((accumulator, currentValue) => accumulator + currentValue, 0); 
console.log(sum) // 10
// The accumulator is the running total, starting at 0.
````

````typescript
// find, some, every
/** These array methods are used for searching or testing elements based on a condition:
 * - `find(callback)`: Returns the *first* element in the array that satisfies the provided testing function. Returns `undefined` if no elements satisfy the test.
 * - `some(callback)`: Tests whether *at least one* element in the array passes the test implemented by the provided function. Returns `true` if it finds an element
 * for which the callback returns a truthy value; otherwise, it returns `false`.
 * - `every(callback)`: Tests whether *all* elements in the array pass the test implemented by the provided function. Returns `true` if the callback returns a truthy
 * value for every element; otherwise, it returns `false`.
 */

const numbers = [10, 20, 30, 40, 50];

const found = numbers.find(num => num > 25);
console.log(found); // Output: 30 (the first element > 25)

const hasLargeNumber = numbers.some(num => num > 45);
console.log(hasLargeNumber); // Output: true (50 is > 45)

const allEven = numbers.every(num => num % 2 == 0);
console.log(allEven);
````

````typescript
// Sorting Arrays
/** The `sort()` method sorts the elements of an array *in place* and returns the sorted array. By default, it sorts elements as strings, which can lead to unexpected results for numbers.
 * To sort numbers correctly or to sort based on custom criteria, you provide a compare function as an argument to `sort()`
 */

//array.sort(compareFunction)

/** The compare function takes two arguments (a, b) and should return:
 *      - A negative value if 'a' should come before 'b'
 *      - A positive value if 'a' should come after 'b'
 *      - 0 if they are considered equal for sorting purposes.
 */

const fruits = ["banana", "apple", "cherry"];
fruits.sort();
console.log(fruits); // Output: ["apple", "banana", "cherry"] (alphabetical)

const numbers = [40, 10, 200, 90];
numbers.sort();
console.log(numbers); // Output: [10, 200, 40, 90] (Incorrect numeric sort)

// Correct numeric sort
numbers.sort(function(a, b) {
    return a - b; // Ascending order
});

console.log(numbers); // Output: [10, 40, 90, 200]

// Descending order
numbers.sort((a, b) => b - a);
console.log(numbers); // Output: [200, 90, 40, 10]
````

````typescript
// String Length and Indexing
/** Strings have a `length` property that returns the number of characters in the string.
 * You can access individual characters in a string using zero-based indexing, similar to arrays, with square brackets [].
 */

let message = "JavaScript";

console.log(message.length); // Output: 10

console.log(message[0]); // Output J
console.log(message[message.length - 1]); // Output: t (last character)

console.log(message[10
]); // Output: undefined (index out of bounds)

// Note that accessing characters with [] is read-only; you cannot change a character in a string this way (strings are
// immutable).
````

````typescript
// Common String Methods
/** Strings have many built-in methods for manipulation and analysis. Some common ones include:
 *      - `toUpperCase()`: Converts the string to uppercase.
 *      - `toLowerCase()`: Converts the string to lowercase.
 *      - `indexOf(substring)`: Returns the index of the first occurrence of a substring, or -1 if not found.
 *      - `lastIndexOf(substring)`: Returns the index of the last occurrence of a substring.
 *      - `slice(startIndex, endIndex)`: Extracts a portion of a string.
 *      - `substring(startIndex, endIndex)`: Similar to slice, with some differences in handling negative indices.
 *      - `replace(searchValue, newValue)`: Replaces the first occurrence of a substring. Use a regular expression with the global
 *         flag ('g') to replace all occurrences.
 *      - `includes(substring)` (ES6): Checks if a string contains a substring, returns true or false.
 *      - `startsWith(substring)` (ES6): Checks if a string starts with a substring.
 *      - `endsWith(substring)` (ES6): Checks if a string ends with a substring.
 */

let text = "Hello World";

console.log(text.toUpperCase()); // Output: HELLO WORLD
console.log(text.toLowerCase()); // Output: hello world
console.log(text.indexOf("World")); // Output: 6
console.log(text.lastIndexOf("Hello")); // Output: 0
console.log(text.slice(0, 5)); // Output: Hello
console.log(text.substring(0, 5)); // Output: Hello
console.log(text.replace("World", "Universe")); // Output: Hello Universe
console.log(text.includes("Universe")); // Output: False
console.log(text.startsWith("Hello", 0)); // Output: True
console.log(text.endsWith("World", 1)); // Output: False
````

````typescript
// Template Literals
/** Template literals (ES6), enclosed in backticks (```), provide a more flexible way to work with strings. They allow for
 *      - Multi-line strings: Strings can span multiple slines without special escape characters.
 *      - Sring interpolation: Embedding expressions (variables, funciton calls, arithmetic) directly within the string using the
 *        `$(expression)` syntax.
 */

let name = "Alice";
let age = 30;

// Using template literal for interpolation
let greeting = `Hello, ${name}! You are ${age} years old.`;
console.log(greeting); // Output: Hello, Alice! You are 30 years old

// Multi-line string
let poem = `
Roses are red,
Violets are blue,
JavaScript is fun,
And so are you!
`;
console.log(poem);

// This is generally preferred over traditional string concatenation with the '+' operator for readability.
````
