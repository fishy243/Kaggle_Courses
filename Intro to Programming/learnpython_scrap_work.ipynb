```python
# Named Tuples
from collections import namedtuple

# Define a named tuple type
# The first argument is the name of the new named tuple type
# The second argument is a string of field names separated by spaces or a list of strings
Point = namedtuple('Point', ['x', 'y'])

# Create an instance of the named tuple
p1 = Point(x=1, y=2)
p2 = Point(3, 4) # You can also use positional arguments

print(p1.x) # Output: 1
print(p2.y) # Output: 4
print(p1[0]) # Output: 1
print(p2[1]) # Output: 4

# namedtuple instances are immutable, just like regular tuples. They inherit from the tuple class, so they
# can be used wherever regular tuples are expected.
# namedtuple is particularly useful for creating simple classes or data structures without the boilerplate
# of a full class definition, improving code readability when dealing with structured data.
```

```python
# Tuple Use Cases
def get_dimensions():
    width = 100
    height = 200
    return width, height # Implicitly returns (width, height)

w, h = get_dimensions() # Tuple unpacking
print(f"Width: {w}, Height: {h}")

coordinate = (40.7128, -74.0060)
color = (255, 0, 0) # Red

location_data = {
    (40.7128, -74.0060): "New York City",
    (34.0522, -118.2437): "Los Angeles"
}
print(location_data[(40.7128, -74.0060)]) # Output: NYC

# Protecting Data Integrity: Using a tuple ensures that the data collection passed around in your program is
# not accidentally modified. 
# Iterating over items that shouldn't be changed: When processing a collection where you only need
# to read the values, a tuple might be slightly more performant than a list. 
# Sequence Unpacking: As seen in tuple packing/unpacking, tuples are fundamental to this feature.
```

```python
# Accessing Values. [] .get()
person = {"name": "Alice", "age": 30, "city": "New York"}

print(person["name"]) # Output: Alice
print(person["age"]) # Output: 30
# print(person["zip_code"]) # Raises KeyError

print(person.get("name")) # Output: Alice
print(person.get("zip_code")) # Output: None (key not found)
print(person.get("zip_code", "Not Available")) # Output: Not Available (provides a default)
print(person.get("age")) # Output: 30

print("name" in person) # Output: True
print("zip_code" in person) # Output: False
```

```python
# Merging Dictionaries
dict1= {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged_dict = {**dict1, **dict2}
print(merged_dict) # Output: {'a': 1, 'b': 3, 'c': 4}

dict_a = {"x": 1, "y": 2}
dict_b = {"y": 20, "z": 30}
merged = {**dict_a, **dict_b}
print(merged) # Output: {'x': 1, 'y': 20, 'z': 30}

# Using the | operator (Merge Operator - Python 3.9+)
# More explicit way to merge dictionaries
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}
merged_dict_new = dict1 | dict2 # Creates a new dictionary
print(merged_dict_new) # Output: {'a': 1, 'b': 3, 'c': 4}

dict1 |= dict2 # Updates dict1 in place (like update())
print(dict1) # Output: {'a': 1, 'b': 3, 'c': 4}

```

```python
# Using dict() Constructor
"""Besides using curly braces {}, the built-in dict()
constructor can be used to create dictionaries in 
several ways. This is particularly useful when creating
dictionaries from other data structures."""

empty_dict = dict()
print(empty_dict) # Output: {}

# Creating from a list of key-value pairs (iterables of 2 items):
"""The argument to dict() can be an iterable (like a list or tuple) where each
element is itself an iterable of exactly two items (key, value)"""

list_of_pairs = [('a', 1), ('b', 2), ('c', 3)]
my_dict_from_list = dict(list_of_pairs)
print(my_dict_from_list) # Output: {'a': 1, 'b': 2, 'c': 3}

tuple_of_pairs = (['x', 10], ['y', 20])
my_dict_from_tuple = dict(tuple_of_pairs)
print(my_dict_from_tuple) # Output: {'x': 10, 'y': 20}

# Creating using keyword arguments:
""""You can pass keyword arguments directly to the dict() constructor, where the keyword
names become the keys (as strings) and the argument values become the dictionary values."""

person = dict(name="Bob", age=25, city="London")
print(person) # Output: {'name': 'Bob', 'age': 25, 'city': 'London'}
# Note: when using keyword arguments, the keys must be valid Python identifiers (e.g, no spaces, cannot start with a number).

# Combining Methods
"""You can combine an iterable of key-value pairs with keyword arguments. The keyword arguments will override any duplicate keys from
the iterable."""

combined_dict = dict([('a', 1), ('b', 2)], b=3, c=4)
print(combined_dict) # Output: {'a': 1, 'b': 3, 'c': 4}
# 'b' from keyword argument overrides the one from the list
```

```python
# Creating Sets
"""A Set in Python is an unordered collection of **unique** and **immutable** elements. Sets are mutable, meaning you can add or
remove elements, but the elements themselves must be immutable. Sets are created using {} or the built-in set() constructor"""

empty_set = set() # Use set() for an empty set! {} creates a dictionary.
my_set = {1, 2, 3, 4, 5}
mixed_set = {"apple", 3.14, True} # Elements must be immutable

# Sets automatically remove duplicates
set_with_duplicates = {1, 2, 2, 3, 1}
print(set_with_duplicates) # Output: {1, 2, 3} (order is not guaranteed)
# Elements in a set must be hashable, which means they must be immutable. Mutable objects like lists or dictionaries cannt be
# elements of a set.
# invalid_set = {[1, 2], 3} # TypeError: cannot use 'list' as a set element (unhashable type: 'list')

my_list = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(my_list)
print(unique_numbers) # Output: {1, 2, 3, 4, 5}

my_string = "programming"
unique_chars = set(my_string)
print(unique_chars) # Output: {'g', 'r', 'a', 'm', 'i', 'p', 'o', 'n'} (order varies)
# Sets are useful for membership testing, removing duplicates, and performing mathematical set operations.
```

```python
# Adding/Removing Elements
"""Sets are mutable, allowing you to add and remove elements after creation.
Adding Elements:
    - set.add(item) : Adds asingle item to the set. If the item is already present, the set remains unchanged (sets only store
    unique elements)."""

my_set = {1, 2, 3}
my_set.add(4)
print(my_set) # Output: {1, 2, 3, 4}

my_set.add(3) # Adding 3 again does nothing
print(my_set) # OutPut: {1, 2, 3, 4}

# set.update(iterable): Adds all elements from an iterable (like a list, tuple, or another set) to the set.

my_set = {1, 2}
my_set.update([3, 4, 2])
print(my_set) # Output: {1, 2, 3, 4}

"""
Removing Elements:
    - set.remove(item) : Removes the specified item from the set. Raises a 'KeyError' if the item is not found."""

my_set = {1, 2, 3, 4}
my_set.remove(3)
print(my_set) # Output: {1, 2, 4}

# my_set.remove(5) # Raises KeyError
# set.discard(item) : Removes the specified item from the set if it is present. Unlike remove(), it does **not** raise an 
# error if the item is not found.

my_set = {1, 2, 3, 4}
my_set.discard(3)
print(my_set) # Output: {1, 2, 4}

my_set.discard(5) # Does nothing if 5 is not found
print(my_set) # Output: {1, 2, 4}

# set.pop() : Removes and returns an arbitrary element from the set. Raises a 'KeyError' if the set is empty. Since sets are
# unordered, there's no guarantee which element will be removed.

my_set = {1, 2, 3}
element = my_set.pop()
print(my_set) # Output: {2, 3} (example output, could be other element removed)
print(element)

# set.clear() : Removes all elements from the set.

my_set = {1, 2, 3}
my_set.clear()
print(my_set) # Output: set()
```

```python
# Set Methods
"""Besides adding and removing elements, sets have methods for performing mathematical set operations and
checking relationships between seets.
Common Set Methods:
    - set.union(other_set, ...) : Returns a new set containing all elements from all sets. Equivalent to the
    | operator."""

set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)
print(union_set) # Output: {1, 2, 3, 4

# set.intersection(other_set, ...) : Returns a new set containing only elements that are present in all sets. 
# Equivalent to the & operator.

set = {1, 2, 3}
set2 = {3, 4, 5}
intersection_set = set1.intersection(set2)
print(intersection_set) # Output: {3}

# set.difference(other_set, ...) : Returns a new set containing elements that are in the first set but not in 
# the other sets. Equivalent to the - operator.

set1 = {1, 2, 3}
set2 = {3, 4, 5}
difference_set = set1.difference(set2)
print(difference_set) # Output: {1, 2}

# set.symmetric_difference(other_set) : Returns a new set containing elements that are in either of the sets 
# but not in both. Equivalent to the ^ operator.

set1 = {1, 2, 3}
set2 = {3, 4, 5}
symmetric_difference_set = set1.symmetric_difference(set2)
print(symmetric_difference_set) # Output: {1, 2, 4, 5}

# set.issubset(other_set) : Returns True if all elements of the set are in the other set. Equivalent to 
# <= operator.

set1 = {1, 2}
set2 = {1, 2, 3, 4}
print(set1.issubset(set2)) # Output: True

# set.issuperset(other_set) : Returns True if all elements of the other set are in the set. Equivalent to
# >= operator.

set1 = {1, 2, 3, 4}
set2 = {1, 2}
print(set1.issuperset(set2)) # Output: True

# set.isdisjoint(other_set) : Returns True if the two sets have no elements in common.

set1 = {1, 2}
set2 = {3, 4}
print(set1.isdisjoint(set2)) # Output: True
# Many set methods have corresponding operator equivalents (e.g., union() vs |, intersection() vs &, 
# difference() vs -, symmetric_difference() vs ^). The operator version are often more concise.
#  Using operators can make the code more concise and readable, especially for simple set operations.
```

```python
# Set Operations
"""Set Operations (with Operators): Python sets support mathematical set operations using special
operators, which are often more concise than the corresponding methods."""

set1 = {1, 2, 3}
set2 = {3, 4, 5}

# Union (|): Combines all unique elements from both sets. Equivalent to set1.union(set2).
union_set = set1 | set2
print(union_set)  # Output: {1, 2, 3, 4, 5}

# Intersection (&): Returns elements common to both sets. Equivalent to set1.intersection(set2).
intersection_set = set1 & set2
print(intersection_set)  # Output: {3}

# Difference (-): Returns elements in set1 that are not in set2. Equivalent to set1.difference(set2).
difference_set = set1 - set2
print(difference_set)  # Output: {1, 2}

difference_set_reversed = set2 - set1
print(difference_set_reversed)  # Output: {4, 5}

# Symmetric Difference (^): Returns elements in either set1 or set2 but not in both. 
# Equivalent to set1.symmetric_difference(set2).
symmetric_difference_set = set1 ^ set2
print(symmetric_difference_set)  # Output: {1, 2, 4, 5}

# Subset (<=) and Proper Subset (<): 
# set1 <= set2 checks if set1 is a subset of set2 (all elements of set1 are in set2). 
# set1 < set2 checks for a proper subset (subset but not equal).
set_a = {1, 2}
set_b = {1, 2, 3}
set_c = {1, 2}
print(set_a <= set_b)  # Output: True (set_a is a subset of set_b)
print(set_a < set_b)   # Output: True (set_a is a proper subset of set_b)
print(set_b < set_a)   # Output: False (set_b is not a proper subset of set_a)
print(set_a <= set_c)  # Output: True (set_a is a subset of set_c)
print(set_a < set_c)   # Output: False (set_a is not a proper subset of set_c, they are equal)

# Superset (>=) and Proper Superset (>):
# set1 >= set2 checks if set1 is a superset of set2 (all elements of set2 are in set1).
# set1 > set2 checks for a proper superset (superset but not equal).
set_a = {1, 2, 3}
set_b = {1, 2}
print(set_a >= set_b)  # Output: True (set_a is a superset of set_b)
print(set_a > set_b)   # Output: True (set_a is a proper superset of set_b)

# These operators provide a clear and efficient way to perform common set operations.
```

```python
# Frozen Sets
"""A frozenset is an immutable version of a Python set. While regular sets are mutable (you can add
or remove elements), frozen sets cannot be changed after creation. 
Frozen sets are created using the frozenset() constructor, usually from an iterable."""

# Creating a frozenset
my_frozenset = frozenset([1, 2, 3, 4, 5])
print(my_frozenset) # Output: frozenset({1, 2, 3, 4, 5})

# Since frozen sets are immutable, they are hashable. This means they can be used in contexts
# where regular sets cannot, such as: As elements within another set. As keys in a dictionary.

# A set of frozensets
set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print(set_of_sets) # Output: {frozenset({1, 2}), frozenset({3, 4})}

# A dictionary with frozenset keys
dict_with_frozenset_keys = {frozenset([1, 2]): "Pair 1", frozenset([3, 4]): "Pair 2"}
print(dict_with_frozenset_keys) # Output: {frozenset({1, 2}): 'Pair 1', frozenset({3, 4}): 'Pair 2'}

# Frozen sets support all the query methods of regular sets (like .union(), .intersection(), 
# .issubset()) and the corresponding operators, but they do not have mnethods that would modify
# the set (like .add(), .remove(), or .update()). 
# Use frozen sets when you need an immutable collection of unique elements.
```

```python
# Checking Membership
"""Checking if an element exists within a set is a very fast operation in Python because sets are implemented using
hash tables. This is one of the primary advantages of using sets for membership tests compared to lists or tuples
(where checking membership might involve scanning the entire sequence)."""

# We use the `in` operator to check for membership:
my_set = {"apple", "banana", "cherry"}

print("banana" in my_set) # Output: True
print("grape" in my_set) # Output: False
print("apple" not in my_set) # Output: False

# The time complexity for checking if an item exists in a set is, on average, O(1)(constant time). This makes sets extremely
# efficient for large collections when you frequently need to perform membership tests.

# Ex: Checking for common elements between two lists efficiently
list1 = [1, 2, 3, 4, 5, ...] # potentially very large
list2 = [4, 5, 6, 7, 8, ...]

# Inefficient way (0(n*m))
# common = [item for item in list1 if item in list2]

# Efficient way using a set (0(n + m))
set2 = set(list2) # Convert list2 to a set (0(m)
common = [item for item in list1 if item in set2] # Check membership in set (O(n))
print(common)

# This efficient membership testing makes sets valuable in algorithms and data processing where checking for the presence of items
# is a frequent operation.
```

```python
# Symmetric Difference
"""The symmetric difference of two sets is a new set containing elements that are in either of the sets, but **not** in both (i.e
elements that are unique to each set relative to the other)"""

# You can perform the symmetric difference using the `.symmetric_difference()` method or the `^` operator.

# Assume you have two sets
set_a = {1, 2, 3}
set_b = {3, 4, 5}

symmetric_diff_method = set_a.symmetric_difference(set_b)
print(symmetric_diff_method) # Output: {1, 2, 4, 5} (order may vary)

symmetric_diff_operator = set_a ^ set_b
print(symmetric_diff_operator) # Output: {1, 2, 4, 5} (")

# The symmetric difference can be thought of as the union minus the intersection: (set_a | set_b) - (set_a & set_b)
# Example Breakdown
# - Elements in `set_a` only: `{1,2}`
# - Elements in `set_b` only: `{4,5}`
# - Elements in both: `{3}`
# - Symmetric difference: `{1,2,4,5}` (unique elements from both)

"""The symmetric difference is useful for finding elements that distinguish two sets.
Difference vs. Symmetric Difference:
    - Difference(`-` or `.difference()`): Returns elements in the first set but *not* in the second.
    - Symmetric Difference (`^` or `.symmetric_difference()`): Returns elements unique to *either* set."""
```

```python
# Set Use Cases
"""Sets are a powerful data structure in Python, particularly useful for tasks involving
uniqueness and membership testing.
    Common Use Cases:
    - Removing Duplicates: Easily got the unique elements from a list or other iterable.
    """

my_list = [1, 2, 2, 3, 4, 4, 5]
unique_elements = list(set(my_list))
print(unique_elements) # Output: [1, 2, 3, 4, 5] (order may vary)

# Membership Testing: Quickly check if an element is present in a large collection.
# This is much faster than using the `in` operator with lists or tuples for large datasets.
allowed_users = {"alice", "bob", "charlie"}
user = "bob"
if user in allowed_users:
    print("Access granted.")

# Performing Mathematical Set Operators: Easily compute unions, intersections, differences, and
# symmetric differences between collections.
attended_class_a = {"Alice", "Bob", "Charlie"}
attended_class_b = {"Charlie", "David", "Eve"}

# Students who attended either class
print(attended_class_a | attended_class_b) # Output: {'Alice', 'Bob', 'Charlie', 'David', 'Eve'}

# Students who attended both classes
print(attended_class_a & attended_class_b) # Output: {'Charlie'}

print(attended_class_a - attended_class_b) # Output: {'Alice', 'Bob'}

"""Checking for Disjoint Sets: Determine if two sets have any elements in common using `.isdisjoint()`.
Finding Differences Between Collections: Quickly find which elements are in one collection but not another.
Sets provide an efficient and clean way to handle collections where the order of elements doesn't matter,
and uniqueness or membership testing is a primary concern."""
```
