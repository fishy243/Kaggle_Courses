```typescript
// else if Chain
if (condition) {
    // code if condition is true
} else if (condition2) {
    // code if condition2 is false AND condition2 is true
} else if (condition3) {
    // code if condition1 and condition2 are false AND condition3 is true
} else {
    // code if none of the above conditions are true (optional)
}
```

```typescript
let score = 85;
if (score >= 90) {
    console.log("Grade: A");
} else if (score >= 80) {
    console.log("Grade: B");
} else if (score >= 70) {
    console.log("Grade: C");
} else {
    console.log("Grade: Below C");
}
```

```typescript
// Logical Operators
let age = 20;
let hasLicense = true;
if (age >= 18 && hasLicense) {
    console.log("You can drive.");
}

let isWeekend = false;
let isHoliday = true;

if (isWeekend || isHoliday) {
    console.log("It's a day off!");
}

let isSunny = false;
if (!isSunny) {
    console.log("It's not sunny today.");
}
```

```typescript
// Ternary Usage

// variable = condition ? value_if_true : value_if_false;

let isLoggedIn = true;
let userStatus = isLoggedIn ? "Online" : "Offline";
    console.log(userStatus); // Output: Online

let price = 100;
let discount = (price > 50) ? price * 0.1 : 0;
    console.log(`Discount: $${discount}`); // Output: Discount: $10
// Avoid nesting ternary operators too deeply, as it can make code difficult to read.
```

```typescript
// Truthy vs Falsy
/**
 * In JavaScript and TypeScript, values can be classified as "truthy" or "falsy" based on how they evaluate 
 * in a boolean context.
 * 
 * Falsy values include:
 * false, 0, -0, "" (empty string), null, undefined, NaN (Not-a-Number)
 * 
 * Truthy values include:
 * true, any non-zero number, any non-empty string ("0", "false", ""), arrays (even empty ones: []),
 * objects (even empty ones: {}), funcitons, symbols, BigInts (except On)
 */

if (0) {
    console.log("This won't print"); // 0 is falsy
}

if ("hello") {
    console.log("This will print"); // "hello" is truthy
}

if ([]) {
    console.log("This will print too"); // Empty array is truthy
}
```

```typescript
// Optional Chaining (?.)
/**
 * Optional chaining (ES11) allows you to safely access nested objects properties or call functions
 * without worrying if an intermediate property or function exists. If a property or function in the chain
 * is null or undefined, the expression short-circuits and evaluates to undefined instead of throwing an error.
 */

// object?.property;
// object>?.['property'];
// object?.method();

const user = {
    name: 'Alice',
    address: {
        street: '123 Main St'
    }
};

console.log(user.address.street); // Output: 123 Main St
console.log(user.address?.city); // Output: undefined (no error)
console.log(user.contact?.email); // Output: undefined (user.contact is undefined, no error)

const user2 = {};
console.log(user2.address?.street); // Output: undefined (user2.address is undefined, no error)
// Calling a method that might not exist
```
